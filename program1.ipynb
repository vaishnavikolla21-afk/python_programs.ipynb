```python
#code for arithmeic operators
x=int(input("Enetr a value:"))
y=int(input("Enter a second value:"))
print(x+y)
print(x-y)
print(x*y)
print(x/y)
print(x%y)
print(x//y)
```

```python
#code for comparision operators
x=int(input("Enetr a value:"))
y=int(input("Enter a second value:"))
print(x==y)
print(x!=y)
print(x<y)
print(x>y)
print(x<=y)
print(x>=y)
```

```python
#code for logic operators
x=int(input("Enetr a value:"))
y=int(input("Enter a second value:"))
print(x>y and x!=y)
print(x<y or x!=y)
print(not(x>y and x!=y))
print(x<y and x!=y)
```

```python
#code for assignment operator
a = int(input("Enter a number: "))

a += 5
print("After += 5 :", a)

a -= 2
print("After -= 2 :", a)

a *= 3
print("After *= 3 :", a)

a //= 2
print("After //= 2 :", a)
```

```python
#code for identity operator
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print("a is b :", a is b)
print("a is c :", a is c)
print("a is not c :", a is not c)
```

```python
#code for membership operator
text = input("Enter a word: ")
ch = input("Enter a character: ")

print(ch, "in text :", ch in text)
print(ch, "not in text :", ch not in text)
```

```python
#code for bitwise operator
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("a & b =", a & b)
print("a | b =", a | b)
print("a ^ b =", a ^ b)
print("~a =", ~a)
print("a << 1 =", a << 1)
print("a >> 1 =", a >> 1)
```
