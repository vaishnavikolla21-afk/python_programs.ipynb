```python
l1=[1,2,3,"hello"]
l2=[22,33,44]
print("concatination:",l1+l2)
print("repition:",l1*3)
print("indexing:",l1[3])
print("slicing:",l1[0:3])
print("membership operator:",100 in l2)
for i in l2:
  print(i)
```

```python
#Built in functions for list
print("sum:",sum(l2))
print("max:",max(l2))
print("min:",min(l2))
print("length:",len(l2))
print("reverse:",list(reversed(l2)))
print("sorted:",sorted(l2))
```

```python
# List Methods

l1 = [1, 2, 3, "hello"]
l2 = [22, 33, 44]
l3 = [11, 22, 33, 44]
l4 = [10, 20, 30]

l1.append("python")
print("After append l1 is:", l1)

l3.extend(l4)
print("After extend l3:", l3)

l2.insert(0, "java")
print("After insert l2:", l2)

print("Count method:", l3.count(30))

l4.pop()
print("Pop method:", l4)

l3.pop(1)
print("Pop method on l3:", l3)

l3.remove(33)
print("Remove method on l3:", l3)

l3.sort()
print("Sort:", l3)

l2.reverse()
print("Reverse l2:", l2)
```
