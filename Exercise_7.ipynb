```python
#set operations
A={1,2,3,4,5}
B={4,5,6,7,8}
print("union:",A|B)
print("intersection:",A&B)
print("difference:",A-B)
print("symmetric difference:",A^B)
```

```python
#set methods
A={1,2,3,4,5}
B={4,5,6,7,8}
print("union:",A.union(B))
print("intersection:",A.intersection(B))
print("difference:",A.difference(B))
print("symmetric_difference:",A.symmetric_difference(B))
A.add(9)
print("after addind A=",A)
A.remove(3)
print("after removing A=",A)
A.pop()
print("after pop A=",A)
A.update([6,7,8])
print("after updating A=",A)
A.discard(5)
print("after dicarding A=",A)
A.clear()
print("after clear A=",A)
```

```python
# write a python program to find no.of duplicate in a list
l=[int(i) for i in input().split()]
s=set(l)
print(len(l)-len(s))
```

```python
# write a python program to find how many numbers having duplicates in a list
l=[int(i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)>1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
print(len(s))
```

```python
#write a python program to find non repeated elements
l=[int(i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)==1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
print(len(s))
```
