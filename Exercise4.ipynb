```python
#program that prints reverse a number using functions
def reverse(n):
  rev=0
  while(n>0):
    rem=n%10
    rev=rev*10+rem
    n=n//10
  return rev
num=int(input("enter a number:"))
res=reverse(num)
print(res)
```
