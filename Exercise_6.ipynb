```python
#operations of tuple
t1=(1,2,3,'a','b')
t2=(5,6,7)
print("concatination:",t1+t2)
print("slicing:",t1[1:5])
print("indexing:",t2[1])
print("repetition:",t1*2)
print("member ship:",3 in t1)
print("iteration")
for i in t1:
  print(i)
```

```python
#built function in tuple
t=(2,21,5,7,9)
print(len(t))
print(max(t))
print(min(t))
print(sum(t))
print(sorted(t))
```

```python
#method in tuples
t1=(1,2,3,'g','v')
print("count:",t1.count('g'))
print("index:",t1.index('v'))
```

```python
#zip function
t=(1,2,3)
t1=('a','b','c')
print(tuple(zip(t,t1)))
```
